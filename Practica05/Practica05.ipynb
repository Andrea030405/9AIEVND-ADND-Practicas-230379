# Un recorrido visual y analítico - DonorsChoose

## Ciencia de Datos para el Bien: DonorsChoose

DonorsChoose es una organización sin fines de lucro con sede en Estados Unidos que permite a las personas donar directamente a proyectos de aulas de escuelas públicas.

Hasta enero de 2018, más de **1 millón de proyectos** habían sido financiados a través de DonorsChoose.org. Además, en el **77 % de las escuelas públicas de Estados Unidos**, al menos un proyecto había sido publicado en esta plataforma.

DonorsChoose compartió un conjunto de datos completo que incluye todos los proyectos realizados en el pasado, junto con los recursos relacionados y los metadatos de **docentes, escuelas, donantes y donaciones**.

Estos datos pueden utilizarse para descubrir patrones e información oculta, así como para desarrollar soluciones que permitan a DonorsChoose crear campañas de correo electrónico más personalizadas, recomendando proyectos específicos a donantes anteriores.

En este cuaderno (Notebook) se realiza un análisis y exploración profunda del conjunto de datos. Además de presentar información relevante sobre DonorsChoose, también funciona como un tutorial para realizar **Análisis Exploratorio de Datos (EDA)** y crear diversas visualizaciones.
